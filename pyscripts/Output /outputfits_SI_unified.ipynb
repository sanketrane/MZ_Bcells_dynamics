```python
# open experimental data file 
df = pd.read_csv('data/new_data_MZB.csv')
df = df.sort_values(by='Age at S1K')
print(df)

df_ont = pd.read_csv('data/New_data/ontogeny_counts.csv')
```

```python
#Add one column in df_Ki67 and df_Nfd for grouping the data based on age at bmt

df['agebin'] = pd.cut(df['Age at BMT'], bins=[0,70, 140], labels=['≤10 weeks', '>10 weeks'])
print(df)
```

```python
sns.set_context("notebook", font_scale=1.3)

ts = np.arange(11.1, 732.1, 0.1)
ts1= np.arange(52, 750, 1)
ts2= np.arange(88, 750, 1)


#plot 4*4 grid 
# Plotting
fig = plt.figure(figsize=(10, 8))
gs = fig.add_gridspec(2, 6)  # Create a GridSpec with 2 rows and 4 columns

# Create the subplots
ax1 = fig.add_subplot(gs[0, 0:3])  # First row, spanning columns 0 to 1
ax2 = fig.add_subplot(gs[0, 3:6])  # First row, spanning columns 2 to 3
# second row has 3 plots based on the age at BMT
ax3 = fig.add_subplot(gs[1, 0:3])  # First row, spanning columns 0 to 1
ax4 = fig.add_subplot(gs[1, 3:6])  # First row, spanning columns 2 to 3
text_only_handle = Line2D([], [], linestyle='none')
# Common title for the whole figure
fig.suptitle('Density-dependent loss model with time-varying influx', fontweight='bold', fontsize=19, y=0.93)
# Common title for the bottom row (ax_3 and ax_4)
fig.text(0.5, 0.425, 'Predicted Ki67+ fraction', ha='center', va='center')


# Plot the Mz counts
sns.scatterplot(ax= ax1, data=df, x='Age at S1K', y='total_MZ', hue='agebin', palette=my_pal, s=50, edgecolor='black',linewidth=0.25)
sns.scatterplot(ax=ax1, data=df_ont, x='age.at.S1K', y='MZ_total',  s=50, color='black', edgecolor='black', linewidth=0.5, legend=False)
ax1.plot(ts, np.mean(Unified_T1.totalMz_pred, axis=0), color='black', linestyle='dashed', linewidth=0.35)
# ax1.plot(ts1, np.mean(SHM_T1.MZ_counts_pred1, axis=0), color = my_pal[0], linestyle='dashed',linewidth=0.25)
# ax1.plot(ts2, np.mean(SHM_T1.MZ_counts_pred2, axis=0), color = my_pal[1], linestyle='dashed',linewidth=0.25)

ax1.plot(ts, np.mean(Unified_T2.totalMz_pred, axis=0), color='black', linewidth=0.35)
# ax1.plot(ts1, np.mean(SHM_T2.MZ_counts_pred1, axis=0), color = my_pal[0], linewidth=0.25)
# ax1.plot(ts2, np.mean(SHM_T2.MZ_counts_pred2, axis=0), color = my_pal[1], linewidth=0.25)
ax1.set_xscale('log')
ax1.set_yscale('log')
ax1.set_ylim(5 * 1e3, 1e7)
ax1.set_xlim(10, 800)
ax1.set_xticks([10, 30, 100, 300, 600])
ax1.set_xticklabels([10, 30, 100, 300, 600])
ax1.set_ylabel('')
ax1.tick_params(axis='both')
ax1.set_xlabel('')
ax1.set_title('MZ B cell counts')
ax1.grid(True, which='both', ls="solid", linewidth=0.25, alpha=0.3)
ax1.legend(title='Age at BMT', frameon=False, fontsize=12, title_fontsize=12)

# Plot normalized donor fraction
sns.scatterplot(ax=ax2, data=df, x='Age at S1K', y='Nfd_MZ', hue='agebin', s=50, edgecolor='black', linewidth=0.25, palette=my_pal, legend=False)
ax2.plot(ts1, np.mean(Unified_T1.Nfd_pred1, axis=0), color=my_pal[0], linewidth=0.35, linestyle='dashed')
ax2.plot(ts2, np.mean(Unified_T1.Nfd_pred2, axis=0), color=my_pal[1], linestyle='dashed', linewidth=0.35, label='T1')
ax2.plot(ts1, np.mean(Unified_T2.Nfd_pred1, axis=0), color=my_pal[0],  linewidth=0.35)
ax2.plot(ts2, np.mean(Unified_T2.Nfd_pred2, axis=0), color=my_pal[1],  linewidth=0.35, label='T2')

# ax2.set_xscale('log')
ax2.set_xlim(40, 800)
ax2.set_xticks([75, 150,  300,  600])
ax2.set_xticklabels([75, 150,  300,  600])
ax2.set_xlim(50, 800)
ax2.set_xscale('log')
ax2.set_xticks([75, 150,  300,  600])
ax2.set_xticklabels([75, 150,  300,  600])
ax2.set_ylabel('')
ax2.set_yticks(np.arange(0, 1.25, 0.25))
ax2.set_yticklabels(['0.0', '', '0.5', '', '1.0'])
ax2.set_xlabel('')
ax2.set_title('Normalized donor fraction')
ax2.axhline(y=1, color='black', linestyle='dashed', linewidth=0.25)
ax2.legend(frameon=False, fontsize=12, title_fontsize=12)
ax2.grid(True, which='both', ls="solid", linewidth=0.5, alpha=0.3)     

#Construct a new dataframe for agebin1 and agebin2
df_Ki67_bin1 = df[df['agebin'] == '≤10 weeks']
df_Ki67_bin2 = df[df['agebin'] == '>10 weeks']
# Plot Proportion of Ki67+ MZ cells
sns.scatterplot(ax=ax3, data=df_Ki67_bin1, x='Age at S1K', y='frac_Ki67+_MZ_host', s=50, edgecolor='black', linewidth=0.25, color='#ab2239')
sns.scatterplot(ax=ax3, data=df_Ki67_bin1, x='Age at S1K', y='frac_Ki67+_MZ_donor', s=50, edgecolor='black', linewidth=0.25, color='#3b9bb3')
ax3.plot(ts1, np.mean(Unified_T1.donor_ki_pred1, axis=0), color= '#3b9bb3',linestyle='dashed', linewidth=0.35)
ax3.plot(ts1, np.mean(Unified_T2.donor_ki_pred1, axis=0), color= '#3b9bb3', linewidth=0.35)
ax3.plot(ts1, np.mean(Unified_T1.host_ki_pred1, axis=0), color = '#ab2239', linewidth=0.35, linestyle='dashed')
ax3.plot(ts1, np.mean(Unified_T2.host_ki_pred1, axis=0), color = '#ab2239', linewidth=0.35)
ax3.set_yticks(np.arange(0, 1.25, 0.25))
ax3.set_yticklabels(['0.0', '', '0.5', '', '1.0'])
ax3.set_xscale('log')
ax3.set_xlim(50, 800)
ax3.set_xticks([75, 150, 300, 600])
ax3.set_xticklabels([75, 150, 300, 600])
ax3.set_title('')
ax3.set_ylabel('')
ax3.set_xlabel(' ')
ax3.legend(handles=[text_only_handle], labels=['Age at BMT\n≤10 wks'], frameon=False, handlelength=0, handletextpad=0, loc='upper right', fontsize=12, title_fontsize=12)
ax3.grid(True, which='both', ls="solid", linewidth=0.5, alpha=0.3)      


# Plot Proportion of Ki67+ MZ cells for age bin2

sns.scatterplot(ax=ax4, data=df_Ki67_bin2, x='Age at S1K', y='frac_Ki67+_MZ_host', s=50, edgecolor='black', linewidth=0.25, color='#ab2239')
sns.scatterplot(ax=ax4, data=df_Ki67_bin2, x='Age at S1K', y='frac_Ki67+_MZ_donor', s=50, edgecolor='black', linewidth=0.25, color='#3b9bb3')
ax4.plot(ts2, np.mean(Unified_T1.donor_ki_pred2, axis=0), color= '#3b9bb3', linestyle='dashed',linewidth=0.35)
ax4.plot(ts2, np.mean(Unified_T2.donor_ki_pred2, axis=0), color= '#3b9bb3',  linewidth=0.35)

ax4.plot(ts2, np.mean(Unified_T1.host_ki_pred2, axis=0), color= '#ab2239', linestyle='dashed',linewidth=0.35)
ax4.plot(ts2, np.mean(Unified_T2.host_ki_pred2, axis=0), color= '#ab2239',  linewidth=0.35)
# ax_4.fill_between(ts2, np.quantile(SHM_T1.host_ki_pred2, 0.025, axis=0), np.quantile(SHM_T1.host_ki_pred2, 0.975, axis=0), color= '#ab2239', alpha=0.05)
# ax_4.fill_between(ts2, np.quantile(SHM_T2.host_ki_pred2, 0.025, axis=0), np.quantile(SHM_T2.host_ki_pred2, 0.975, axis=0), color= '#ab2239', alpha=0.05)
# ax_4.fill_between(ts2, np.quantile(SHM_FO.host_ki_pred2, 0.025, axis=0), np.quantile(SHM_FO.host_ki_pred2, 0.975, axis=0), color= '#ab2239', alpha=0.05)
# ax_4.fill_between(ts2, np.quantile(SHM_T2MZP.host_ki_pred2, 0.025, axis=0), np.quantile(SHM_T2MZP.host_ki_pred2, 0.975, axis=0), color= '#ab2239', alpha=0.05)
ax4.set_ylim(0, 1.1)
ax4.set_yticks(np.arange(0, 1.25, 0.25))
ax4.set_yticklabels(['', '', '', '', ''])
ax4.set_xscale('log')
ax4.set_xlim(75, 800)
ax4.set_xticks([75, 150, 300, 600])
ax4.set_xticklabels([75, 150, 300, 600])
ax4.set_title('')
ax4.set_ylabel('')
ax4.set_xlabel('')
donor_handle = Line2D([], [], color='#3b9bb3', marker='o', linestyle='none', markersize=5, label='Donor')
host_handle  = Line2D([], [], color='#ab2239', marker='o', linestyle='none', markersize=5, label='Host')
text_handle  = Line2D([], [], linestyle='none', label='Age at BMT\n$>$10 wks')

ax4.legend(
    handles=[donor_handle, host_handle, text_handle],
    frameon=False,
    handlelength=1.8,
    handletextpad=0.6,fontsize=12, title_fontsize=12
)

ax4.grid(True, which='both', ls="solid", linewidth=0.5, alpha=0.3)

# Axis border width and tick styling for all panels
for ax in [ax1, ax2, ax3, ax4]:
    for spine in ax.spines.values():
        spine.set_linewidth(0.2)   # try 0.6 / 0.8 / 1.0

    # Tick mark width
    ax.tick_params(axis='both', which='both', width=0.8)
    ax.grid(True, axis="x", ls="-.", color='silver', linewidth=0.25)
    ax.grid(True, axis="y", ls="--", color='silver', linewidth=0.25)
    ax.set_xlabel('')
    ax.tick_params(axis="both")
    # tick mark size and thickness
    ax.tick_params(axis="both", which="major", length=3, width=1)
    ax.tick_params(axis="both", which="minor", length=1.5, width=0.25)


plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.savefig('DDM_varyinflux_lossdensity_unified_Supplementary.pdf', dpi=300)
plt.show()

#Save the figure
# plt.savefig('/home/apoorva/Desktop/work/MZ_analysis/figures/new_output/TDM_T1_varyinflux.png', dpi=350, bbox_inches='tight')

```
