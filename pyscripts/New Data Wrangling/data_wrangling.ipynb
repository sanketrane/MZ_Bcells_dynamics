```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# python font no dingbattes
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42
```

```python
# load the data
data_T1 = pd.read_csv('datafiles/T1_data_modified.csv')
data_T2 = pd.read_csv('datafiles/T2_data_modified.csv')
data_T2MZP = pd.read_csv('datafiles/T2_MZP_data_modified.csv')
data_FM = pd.read_csv('datafiles/FM_data_modified.csv')
data_MZ = pd.read_csv('datafiles/MZ_data_modified.csv')

# combine the data with a key column
data_T1['subset'] = 'T1'
data_T2['subset'] = 'T2'
data_T2MZP['subset'] = 'T2MZP'
data_FM['subset'] = 'FM'
data_MZ['subset'] = 'MZ'
data_T1['group'] = 'Transitional'
data_T2['group'] = 'Transitional'
data_T2MZP['group'] = 'Transitional'
data_FM['group'] = 'Mature'
data_MZ['group'] = 'Mature'
data = pd.concat([data_T1, data_T2, data_T2MZP, data_FM, data_MZ], ignore_index=True)
data['subset'] = data['subset'].astype('category')
data['group'] = data['group'].astype('category')

data.head()
```

```python
data['ratio_kifrac_host_to_donor'] = data['host_kifrac'] / data['donor_kifrac']
```

```python
# plot the data as subplots for each group
g = sns.FacetGrid(data, col='group', col_wrap=2, height=4, sharey=True)
g.set_titles(col_template='') # remove facet titles
g.map_dataframe(sns.scatterplot, x='days_post_BMT', y='ratio_kifrac_donor_to_host', palette='Paired', hue='subset', s=40)
g.set(xscale='log')
g.set(xticks=[10, 30, 100, 300], xticklabels=[10, 30, 100, 300])
g.set(ylim=(0.25, 10))
g.set(yscale='log')
g.set(yticks=[0.25, 0.5, 1, 2.5, 5, 10], yticklabels=[0.25, 0.5, 1, 2.5, 5, 10])
g.set_axis_labels('Days Post BMT', 'Ratio of % Ki67+ Donor to Host')
g.add_legend(title=None)
for ax in g.axes.flat:
    ax.grid(True, axis="y", ls="-")
# save the figure
#g.savefig('plots/ratio_kifrac_donor_to_host.pdf', dpi=300, bbox_inches='tight')
```

```python
# plot kifrac as scatterplot
plt.figure(figsize=(6,4))
sns.scatterplot(data=data, x='days_post_BMT', y='ratio_kifrac_donor_to_host', hue='subset', s=40, palette='Paired')
plt.xscale('log')
plt.xticks([10, 30, 100, 300], [10, 30, 100, 300])
plt.yscale('log')
plt.yticks([0.25, 0.5, 1, 2.5, 5, 10], [0.25, 0.5, 1, 2.5, 5, 10])
plt.xlabel('Days Post BMT')
plt.ylabel('Ratio of Ki-Frac (Donor to Host)')
plt.ylim(0.25, 10)
plt.grid(True, axis="y", ls="-")
plt.legend(title=None)
plt.tight_layout()
# save the figure
plt.savefig('plots/kifrac_compare.pdf', dpi=300, bbox_inches='tight')
```

```python
# plot fd as subplots for each group
g = sns.FacetGrid(data, col='group', col_wrap=2, height=4, sharey=True)
g.set_titles(col_template='') # remove facet titles
g.map_dataframe(sns.scatterplot, x='days_post_BMT', y='Nfd', palette='Paired', hue='subset', s=40)
g.set(xscale='log')
g.set(xticks=[10, 30, 100, 300], xticklabels=[10, 30, 100, 300])
g.set(ylim=(0, 1))
g.set(yticks=[0, 0.2, 0.4, 0.6, 0.8, 1], yticklabels=[0, 0.2, 0.4, 0.6, 0.8, 1])
g.set_axis_labels('Days Post BMT', 'Fraction Donor Ki67+ Cells')
g.add_legend(title=None)
for ax in g.axes.flat:
    ax.grid(True, axis="y", ls="-")
# save the figure
g.savefig('plots/fraction_donor_kifrac.pdf', dpi=300, bbox_inches='tight')
```

```python
# plot nfd as scatterplot
plt.figure(figsize=(6, 4))
sns.scatterplot(data=data, x='days_post_BMT', y='Nfd', hue='subset', palette='Paired', s=40)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Days Post BMT')
plt.ylabel('Nfd')
plt.xticks([10, 30, 100, 300], [10, 30, 100, 300])
plt.yticks([0.01, 0.03, 0.1, 0.3, 1], [0.01, 0.03, 0.1, 0.3, 1])
plt.ylim(0.01, 1.02)
plt.grid(True, axis="y", ls="-")
plt.legend(title=None)
plt.tight_layout()
# save the figure
plt.savefig('plots/Nfd_compar_logY.pdf', dpi=300, bbox_inches='tight')
```

```python
# create color gradient using two colors
from matplotlib.colors import LinearSegmentedColormap
my_cmap = LinearSegmentedColormap.from_list("mycmap", ["#f99f1b", "#c46c4c", "#782f6c", "#512975", "#282a73"])
```

```python
# plot mz counts
plt.figure(figsize=(6, 4))
sns.scatterplot(data=data_MZ, x='days_post_BMT', y='counts', hue='age_at_BMT', palette=my_cmap, s=75, edgecolor='black', linewidth=0.5)
plt.title('MZ B cell numbers')
plt.xlabel('Days Post BMT')
plt.ylabel('')
plt.yscale('log')
plt.ylim(1e5, 1e7)
# y axis ticks for log scale
yticks = np.concatenate([np.arange(1, 10, 1)*1e5, np.arange(1, 11, 1)*1e6])
plt.yticks(yticks)
plt.xscale('log')
plt.xticks([10, 30, 100, 300, 600], [10, 30, 100, 300, 600])
#plt.grid(True, axis="y", ls="-")
plt.tight_layout()
# no legend for points
plt.legend([],[], frameon=False)
# save the figure
# #  plot legend as colorbar
norm = plt.Normalize(data_MZ['age_at_BMT'].min()/7, data_MZ['age_at_BMT'].max()/7)
sm = plt.cm.ScalarMappable(cmap=my_cmap, norm=norm)
sm.set_array([])
ax = plt.gca()
cbar = plt.colorbar(sm, ax=ax)
cbar.set_label('Age at BMT (weeks)')
plt.savefig('plots/MZ_counts.pdf', dpi=300, bbox_inches='tight')
```

```python
# plot nfd in MZ B cells as scatterplot 
plt.figure(figsize=(6, 4))
sns.scatterplot(data=data_MZ, x='days_post_BMT', y='Nfd', hue='age_at_BMT', palette=my_cmap, s=75, edgecolor='black', linewidth=0.5)
plt.xscale('log') 
plt.xlabel('Days Post BMT')
plt.title('Donor fraction normalized to chimerism in T1')
plt.ylabel('')
plt.xticks([10, 30, 100, 300, 600], [10, 30, 100, 300, 600])
plt.yticks([0.0, 0.25, 0.5, 0.75, 1.0], [0.0, 0.25, 0.5, 0.75, 1.0])
plt.axhline(y=1.0, color='darkred', linestyle='--', linewidth=1)
#plt.grid(True, axis="y", ls="-")
plt.tight_layout()
# no legend for points
plt.legend([],[], frameon=False)
# save the figure
#  plot legend as colorbar
sm.set_array([])
ax = plt.gca()
cbar = plt.colorbar(sm, ax=ax)
cbar.set_label('Age at BMT (weeks)')
plt.savefig('plots/MZ_Nfd.pdf', dpi=300, bbox_inches='tight')
```
