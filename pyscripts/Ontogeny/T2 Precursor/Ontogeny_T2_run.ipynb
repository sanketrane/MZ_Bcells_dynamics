```python
import os
from cmdstanpy import CmdStanModel

```

```python
#Change working directory to the location of the script
os.chdir('/home/apoorva/Desktop/MZB-new-analysis')
```

```python
stan_file = os.path.join('stan_models/Ontogeny/DDM Influx/T2 Precursor/DDdivision_dependent_constantphi.stan')
```

```python
# instantiate the model object
model = CmdStanModel(stan_file=stan_file, cpp_options={"STAN_THREADS": True})
```

```python
# inspect model object

print(model)
```

```python

# inspect compiled model

print(model.exe_info())
```

```python
#specify data file

data_file = os.path.join('data/MZcounts_ontogeny.json')
# init_dict = {'rho': 0.024, 'delta': 0.035 , 'M_stable': ,'sigma': [1]}

#model fit 
fit = model.sample(
    data=data_file, output_dir='output',show_console=True, 
    # fixed_param= True, inits =init_dict ,
    iter_sampling= 2000, iter_warmup= 500, chains=5, parallel_chains=5, threads_per_chain=5
    # max_treedepth= 20, adapt_delta = 0.99
)
# Print summary of the fit
print(fit.summary())

# Diagnose the fit
diagnostic_output = fit.diagnose()
print(diagnostic_output)
```

```python
#save output to csv file

# fit.save_csvfiles('/home/apoorva/Desktop/work/MZ_analysis/data/fit_T1.csv')

import pickle as pkl

with open('data/New output (pkl)/Ontogeny Output/DDM/T2 Precursor/DDM_T2_Division_dependent_constantphi.pkl', 'wb') as f:
    pkl.dump(fit, f)
```
