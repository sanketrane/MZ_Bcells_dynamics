```python
import os
from cmdstanpy import CmdStanModel

```

```python
# Specify Stan program file

stan_file = os.path.join('/home/apoorva/Desktop/MZB-new-analysis/stan_models/2Agebins_steadystate/Constant Birth-Loss/Constant_birth_lossT2.stan')
```

```python
# instantiate the model object

model = CmdStanModel(stan_file=stan_file, cpp_options={"STAN_THREADS": True})
```

```python
# inspect model object

print(model)
```

```python

# inspect compiled model

print(model.exe_info())
```

```python
#specify data file

data_file = os.path.join('/home/apoorva/Desktop/MZB-new-analysis/data/new_MZdata_2bins.json')
# init_dict = {'y0_Log': 11, 'kappa_0': 0.15, 'rho': 0.024, 'phi': 0.019 , 'sigma': [1,1,1,1]}

#model fit with fixed parameters
#model fit with fixed parameters
fit = model.sample(
    data=data_file, output_dir='/home/apoorva/Desktop/MZB-new-analysis/output',show_console=True, 
    # fixed_param= True, inits =init_dict ,
    iter_sampling= 1000, iter_warmup= 500, chains=5, parallel_chains=5, threads_per_chain=5
    # max_treedepth= 20, adapt_delta = 0.99
)
# Print summary of the fit
print(fit.summary())

# Diagnose the fit
diagnostic_output = fit.diagnose()
print(diagnostic_output)
```

```python
#save output to csv file

# fit.save_csvfiles('/home/apoorva/Desktop/work/MZ_analysis/data/fit_T1.csv')

import pickle as pkl

with open('/home/apoorva/Desktop/MZB-new-analysis/data/New output (pkl)/Steady State/SHM/SHM_T2.pkl', 'wb') as f:
    pkl.dump(fit, f)
```
