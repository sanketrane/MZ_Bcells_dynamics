```python
import os
from cmdstanpy import CmdStanModel

```

```python
# Specify Stan program file

stan_file = os.path.join('/home/apoorva/Desktop/work/MZ_analysis/stan_models/new_models/Steady_State/2_AgeBins/Unified_Model/Linear.neutral.feedback.T1fitall_PrecursorDriven.stan')
```

```python
# instantiate the model object

model = CmdStanModel(stan_file=stan_file, cpp_options={"STAN_THREADS": True})
```

```python
# inspect model object

print(model)
```

```python

# inspect compiled model

print(model.exe_info())
```

```python
#specify data file

data_file = os.path.join('/home/apoorva/Desktop/work/MZ_analysis/data/newdata_mz_allage_2agebins.json')
# init_dict = {'y0_Log': 11, 'kappa_0': 0.15, 'rho': 0.024, 'phi': 0.019 , 'sigma': [1,1,1,1]}

#model fit with fixed parameters
fit = model.sample(
    data=data_file, output_dir='/home/apoorva/Desktop/work/MZ_analysis/output',show_console=True, 
    # fixed_param= True, inits =init_dict ,
    iter_sampling= 500, iter_warmup= 200, chains=4, parallel_chains=4, threads_per_chain=10
    # max_treedepth= 20, adapt_delta = 0.99
)
# Print summary of the fit
print(fit.summary())

# Diagnose the fit
diagnostic_output = fit.diagnose()
print(diagnostic_output)
```

```python
import pickle as pkl

with open('/home/apoorva/Desktop/work/MZ_analysis/data/New output (pkl)/Steady State/Unified_model/Feedback_T1_PrecursorDriven.pkl', 'wb') as f:
    pkl.dump(fit, f)
```
